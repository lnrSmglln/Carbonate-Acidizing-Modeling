### Химия

Рассмотрим реакцию соляной кислоты с известняком (кальцитом):

$$
\begin{gathered}
\text{CaCO}_3 + 2 \text{HCl} \rightarrow \text{CaCl}_2 + \text{H}_2\text{O} + \text{CO}_2
\\
\bold{r}ock \hspace{0.75cm} \bold{a}cid \hspace{0.75cm} \bold{s}alt \hspace{0.55cm} \bold{w}ater \hspace{0.25cm} \bold{g}as
\end{gathered}
$$

В массовом виде относительно кальцита уравнение реакции выглядит так:
$$
\begin{gathered}
\text{CaCO}_3 + 0.729 \, \text{HCl} \rightarrow 1.109 \, \text{CaCl}_2 + 0.18 \, \text{H}_2\text{O} + 0.44 \, \text{CO}_2
\end{gathered}
$$

Относительно кислоты:
$$
\begin{gathered}
1.373 \, \text{CaCO}_3 + \text{HCl} \rightarrow 1.523 \, \text{CaCl}_2 + 0.247 \, \text{H}_2\text{O} + 0.604 \, \text{CO}_2
\end{gathered}
$$

Массовые стехиометрические коэффициенты:

- $\chi_a = 0.728579$
- $\chi_s = 1.108884$
- $\chi_w = 0.179992$
- $\chi_g = 0.439703$

Предполагается, что смачивающая фаза, в основном состоящая из воды, несет в себе кислоту, полностью растворимые соль и углекислый газ

Кинетика реакции описывается формулой:

$$
J = k_a S (1 - \phi) \frac{M_r}{2} S_s \rho_a^m \frac{(c_{a} - c_{\text{eq}})^m}{M_{a}^m}
$$

- $M_r = 0.100088$ кг/моль
- $M_a = 0.036461$ кг/моль

### МСС

#### Уравнения неразрывности

$$
\frac{\partial \rho_A}{\partial t} + \nabla \cdot (\rho_A \vec{v}_A) = \chi_A J
$$

В одномерном случае получим систему уравнений

- для породы:

$$
\frac{\partial}{\partial t} \left[ \rho_r (1 -\phi) \right] = - J
$$

- для воды и растворенных компонентов (кислоты, соли, газа и полимера) $\left( c_a + c_s + c_g + c_p = 1 - c_w \right)$

$$
\begin{gathered}
\frac{\partial}{\partial t} \left[ \rho_a \phi S c_a \right] +
\frac{\partial}{\partial x} \left[ \rho_a c_a \phi S v_w \right] =
- \chi_a J
\\
\frac{\partial}{\partial t} \left[ \rho_s \phi S c_s \right] +
\frac{\partial}{\partial x} \left[ \rho_s c_s \phi S v_w \right] =
\chi_s J
\\
\frac{\partial}{\partial t} \left[ \rho_w \phi S c_w \right] +
\frac{\partial}{\partial x} \left[ \rho_w с_w \phi S v_w \right] =
\chi_w J
\\
\frac{\partial}{\partial t} \left[ \rho_g \phi S c_g \right] +
\frac{\partial}{\partial x} \left[ \rho_g c_g \phi S v_w \right] =
\chi_g J
\\
\frac{\partial}{\partial t} \left[ \rho_p \phi S c_p \right] +
\frac{\partial}{\partial x} \left[ \rho_p c_p \phi S v_w \right] =
0
\end{gathered}
$$

- для нефти:

$$
\frac{\partial}{\partial t} \left[ \rho_o \phi (1 - S) \right] + 
\frac{\partial}{\partial x} \left[ \rho_o \phi (1 - S) v_o \right] = 0
$$

Скорость фильтрации по определению:
$$
\vec{W}_w = \phi S \vec{v}_w
\hspace{1cm}
\vec{W}_o = \phi (1 - S) \vec{v}_o
$$

Из з. Дарси скорости фильтрации:

$$
\begin{gathered}
\vec{W}_w = -k \frac{k_w}{\mu_w(c_p, ...)} \nabla P
\hspace{1cm}
\vec{W}_o = -k \frac{k_o}{\mu_o} \nabla P
\\
\vec{W}_o = \vec{W}_w \frac{k_o}{k_w} \frac{\mu_w}{\mu_o}
\\
\vec{W}_o + \vec{W}_w = \vec{W}_w \left( 1 + \frac{k_o}{k_w} \frac{\mu_w}{\mu_o} \right)
\\
\frac{\vec{W}_w}{\vec{W}_o + \vec{W}_w} = \frac{1}{1 + \frac{k_o}{k_w} \frac{\mu_w}{\mu_o}} = b(S, c_p, ...)
\\
\vec{W}_w = \vec{W} b
\hspace{1cm}
\vec{W}_o = \vec{W} ( 1 - b)
\end{gathered}
$$

Здесь пренебрегаются капиллярное давление, влияние кол-ва растворенного $\text{CO}_a$ в воде на $\mu_w$

Таким образом:
- для породы:
  
$$
\frac{\partial}{\partial t} \left[ \rho_r (1 -\phi) \right] = - J
$$

- для воды и растворенных компонентов
  
$$
\begin{gathered}
\frac{\partial}{\partial t} \left[ \rho_a \phi S c_a \right] +
\nabla \cdot \left[ \rho_a c_a \vec{W}b \right] =
- \chi_a J
\\
\frac{\partial}{\partial t} \left[ \rho_s \phi S c_s \right] +
\nabla \cdot \left[ \rho_s c_s \vec{W}b \right] =
\chi_s J
\\
\frac{\partial}{\partial t} \left[ \rho_w \phi S c_w \right] +
\nabla \cdot \left[ \rho_w c_w \vec{W}b \right] =
\chi_w J
\\
\frac{\partial}{\partial t} \left[ \rho_g \phi S c_g \right] +
\nabla \cdot \left[ \rho_g c_g \vec{W}b \right] =
\chi_g J
\\
\frac{\partial}{\partial t} \left[ \rho_p \phi S c_p \right] +
\nabla \cdot \left[ \rho_p c_p \vec{W}b \right] =
0
\end{gathered}
$$

- для нефти:

$$
\frac{\partial}{\partial t} \left[ \rho_o \phi (1 - S) \right] + 
\nabla \cdot \left[ \rho_o \vec{W} (1-b) \right] = 0
$$

Пусть жидкости и порода несжимаемы, тогда:

- для породы:
  
$$
\frac{\partial \phi}{\partial t} = \frac{1}{\rho_r} J
$$

- для воды и растворенных компонентов

$$
\begin{gathered}
\frac{\partial}{\partial t} \left[ \phi S c_a \right] +
\nabla \cdot \left[  c_a Wb \right] =
- \frac{\chi_a}{\rho_a} J
\\
\frac{\partial}{\partial t} \left[ \phi S c_s \right] +
\nabla \cdot \left[ c_s Wb \right] =
\frac{\chi_s}{\rho_s} J
\\
\frac{\partial}{\partial t} \left[ \phi S c_w \right] +
\nabla \cdot \left[ c_w Wb \right] =
\frac{\chi_w}{\rho_w} J
\\
\frac{\partial}{\partial t} \left[ \phi S c_g \right] +
\nabla \cdot \left[ c_g Wb \right] =
\frac{\chi_g}{\rho_g} J
\\
\frac{\partial}{\partial t} \left[ \phi S c_p \right] +
\nabla \cdot \left[ c_p Wb \right] =
0
\end{gathered}
$$

- для нефти:

$$
\frac{\partial}{\partial t} \left[ \phi (1 - S) \right] + 
\nabla \cdot \left[ W (1-b) \right] = 0
$$

Сложив уравнения для воды и компонентов, получим:
$$
\frac{\partial \phi S}{\partial t} +
\nabla \cdot [\vec{W}b ] =
J \left( -\frac{\chi_a}{\rho_a} +\frac{\chi_s}{\rho_s} + \frac{\chi_w}{\rho_w} + \frac{\chi_g}{\rho_g} \right)
$$

Складывая это уравнение с уравнением неразрывности для нефти получим:
$$
\frac{\partial \phi}{\partial t} +
\nabla \cdot \vec{W} =
J \left( -\frac{\chi_a}{\rho_a} +\frac{\chi_s}{\rho_s} + \frac{\chi_w}{\rho_w} + \frac{\chi_g}{\rho_g} \right)
$$
Подставляя вместо $\frac{\partial \phi}{\partial t}$ правую часть уравнения неразрывности для породы:
$$
\nabla \cdot \vec{W} =
J \left( -\frac{1}{\rho_r} -\frac{\chi_a}{\rho_a} +\frac{\chi_s}{\rho_s} + \frac{\chi_w}{\rho_w} + \frac{\chi_g}{\rho_g} \right)
$$

В цилиндрических координат дивергенция примет вид:

$$\nabla \cdot \vec{W} = {1 \over r }{\partial \left(r W_{r }\right) \over \partial r } + {1 \over r }{\partial W_{\varphi } \over \partial \varphi }+{\partial W_{z} \over \partial z} $$

Как правило свойства пласта по осям $\varphi$ и $z$ отсутствуют, задача сводится к одномерной радиальной:

$$\nabla \cdot \vec{W} = {1 \over r }{\partial \left(r W_{r }\right) \over \partial r }$$

$$
{1 \over r }{\partial \left(r W\right) \over \partial r } =
J \left( -\frac{1}{\rho_r} -\frac{\chi_a}{\rho_a} +\frac{\chi_s}{\rho_s} + \frac{\chi_w}{\rho_w} + \frac{\chi_g}{\rho_g} \right)

### Дискретизация

Дискретизируем явно разностями против потока полученную систему:

$$
\begin{gathered}
J_i^n = k_a S_i^n (1 - \phi_i^n) \frac{M_1}{2} S_s \rho_a^m \frac{({c_a}_i^n - c_{\text{eq}})^m}{M_a^m}
\\
\frac{W_i^n - W_{i-1}^n}{h} =
J_i^n \left( -\frac{1}{\rho_1} -\frac{\chi_a}{\rho_a} +\frac{\chi_s}{\rho_s} + \frac{\chi_w}{\rho_w} + \frac{\chi_g}{\rho_g} \right)
\\
\frac{\phi_i^{n+1} - \phi_i^n}{\tau} = \frac{1}{\rho_1} J_i^n
\\
\frac{\phi_i^{n+1} S_i^{n+1} - \phi_i^n S_i^n}{\tau} +
\frac{W_i^n b_i^n - W_{i-1}^n b_{i-1}^n}{h} =
J_i^n \left( -\frac{\chi_a}{\rho_a} +\frac{\chi_s}{\rho_s} + \frac{\chi_w}{\rho_w} + \frac{\chi_g}{\rho_g} \right)
\\
\frac{\phi_i^{n+1} S_i^{n+1} {c_a}_i^{n+1} - \phi_i^n S_i^n {c_a}_i^n}{\tau} +
\frac{ {c_a}_i^n W_i^n b_i^n - {c_a}_{i-1}^n W_{i-1}^n b_{i-1}^n  }{h} =
- \frac{\chi_a}{\rho_a} J_i^n
\\
\frac{\phi_i^{n+1} S_i^{n+1} {c_s}_i^{n+1} - \phi_i^n S_i^n {c_s}_i^n}{\tau} +
\frac{ {c_s}_i^n W_i^n b_i^n - {c_s}_{i-1}^n W_{i-1}^n b_{i-1}^n  }{h} =
\frac{\chi_s}{\rho_s} J_i^n
\\
\frac{\phi_i^{n+1} S_i^{n+1} {c_g}_i^{n+1} - \phi_i^n S_i^n {c_g}_i^n}{\tau} +
\frac{ {c_g}_i^n W_i^n b_i^n - {c_g}_{i-1}^n W_{i-1}^n b_{i-1}^n  }{h} =
\frac{\chi_g}{\rho_g} J_i^n
\end{gathered}
$$

Алгоритм расчета:

1. Расчет скорости реакции

$$
J_i^n = k_a S_i^n (1 - \phi_i^n) \frac{M_1}{2} S_s \rho_a^m \frac{({c_a}_i^n - c_{\text{eq}})^m}{M_a^m}
$$

2. Расчет скорости фильтрации

$$
W_i^n = W_{i-1}^n +
h J_i^n \left( -\frac{1}{\rho_r} -\frac{\chi_a}{\rho_a} +\frac{\chi_s}{\rho_s} + \frac{\chi_w}{\rho_w} + \frac{\chi_g}{\rho_g} \right)
$$

3.  Расчет изменения пористости

$$
\phi_i^{n+1} = \phi_i^n + \frac{\tau}{\rho_r} J_i^n
$$

4. Расчет водонасыщенности

$$
S_i^{n+1}
 = \frac{1}{\phi_i^{n+1}} \left\{ \phi_i^n S_i^n +
\tau \left[ J_i^n \left( -\frac{\chi_a}{\rho_a} +\frac{\chi_s}{\rho_s} + \frac{\chi_w}{\rho_w} + \frac{\chi_g}{\rho_g} \right)
- \frac{W_i^n b_i^n - W_{i-1}^n b_{i-1}^n}{h} \right] \right\}
$$

5. Расчет концентраций
   
$$
\begin{gathered}
{c_a}_i^{n+1} 
= \frac{1}{\phi_i^{n+1} S_i^{n+1}}
\left[ \phi_i^n S_i^n {c_a}_i^n - \tau \left(\frac{\chi_a}{\rho_a} J_i^n + \frac{ {c_a}_i^n W_i^n b_i^n - {c_a}_{i-1}^n W_{i-1}^n b_{i-1}^n }{h} \right) \right]
\\

{c_s}_i^{n+1} 
= \frac{1}{\phi_i^{n+1} S_i^{n+1}}
\left[ \phi_i^n S_i^n {c_s}_i^n + \tau \left( \frac{\chi_s}{\rho_s} J_i^n - \frac{ {c_s}_i^n W_i^n b_i^n - {c_s}_{i-1}^n W_{i-1}^n b_{i-1}^n }{h} \right) \right]
\\

{c_g}_i^{n+1} 
= \frac{1}{\phi_i^{n+1} S_i^{n+1}}
\left[ \phi_i^n S_i^n {c_g}_i^n + \tau \left(\frac{\chi_g}{\rho_g} J_i^n - \frac{ {c_g}_i^n W_i^n b_i^n - {c_g}_{i-1}^n W_{i-1}^n b_{i-1}^n }{h} \right) \right]
\end{gathered}
$$

$$
\tau ≤ {h \over 2 W}
$$

#### Дополнительный ограничитель на шаг по времени


$$
\begin{gathered}
\phi_i^{n+1} = \phi_i^n + \frac{\tau}{\rho_r} J_i^n
\\
\phi_i^{n+1} ≤ 1
\\
\tau ≤ \frac{\rho_r}{J_i^n}(1 - \phi^n_i)
\\
\tau ≤ \frac{2 \rho_r M_a^m}{k_a M_r S_s \rho_a^m ({c_a}_{\text{input}} - c_{\text{eq}})^m}
\end{gathered}
$$

```python
# Кол-во узлов сетки
num_of_cell_points = 25
num_of_grid_points = num_of_cell_points

# Геометрия
height = 5
r_well = 0.1
r_res = 3

# grid = np.linspace(r_well, r_res, num_of_cell_points+1, endpoint=True)[:-1]
grid = np.logspace( np.log10(r_well), np.log10(r_res), num_of_cell_points+1, endpoint=True)[1:]

# Хар-ки породы
initial_permeability = 1e-14
initial_porosity = 0.2

rho1 = 2740
M1 = 0.100088

# Хар-ки флюидов
viscosity_water = 3 * 10**(-3)
viscosity_oil = 4 * 10**(-3)
M2 = 0.036461
rho2 = 1200
rho3 = 2700
rho4 = 1000
rho5 = 1167

alpha_rock = 50.045 / 1000
alpha_acid = 36.46 / 1000
alpha_salt = 55.49 / 1000
alpha_water = 9.01 / 1000
alpha_co2 = 22.005 / 1000

# Расход
# flow = 1e-4
# volume = 1e-3

# filtration_velocity = 1.75 / 1000
# filtration_velocity = 1.75 / 100
# filtration_velocity = 3.5 / 100
# filtration_velocity = 1.75 / 10
# filtration_velocity = 3.5 / 10
# filtration_velocity = 1.75 / 1
# filtration_velocity = 1.75 / 200
filtration_velocity = (initial_area_per_volume * k_a * np.power(conc_left * rho2 / M2, m-1) / 2) / 0.001
inject_pore_volume = 10
volume = inject_pore_volume * initial_porosity * np.pi * (r_res**2 - r_well**2) * height

pressure_right = 1e7

# Насыщенность
saturation_left = 1 - S_ORW
saturation_initial = S_WIR
# saturation_initial = 0.0
conc_left = 0.15

# Параметры реакции
k_a = 1.4e-5
m = 1.202
# m = 0.42
initial_area_per_volume = 50000
c_eq = 0.0
beta = 0.84

flow = filtration_velocity * 2 * np.pi * r_well * height


# Da = r_res * initial_area_per_volume * k_a * np.power(conc_left * rho2 / M2, m) * \
#      (M2 / 2) / \
#      filtration_velocity
# print(f"Q = {flow * 1e3 * 60:.1f} l/min\n1/Da = {1/Da:.1e}")


Da = initial_area_per_volume * k_a * np.power(conc_left * rho2 / M2, m-1) / 2 / filtration_velocity

print(f"Q = {flow * 1e3 * 60:.1f} l/min\n1/Da = {1/Da:.1e}")
```

### Инициализация

```python
# оценка времени расчета
total_time = volume / flow
# total_time = length * s_right / b(s_right, viscosity_water, viscosity_oil) / velocity * initial_porosity * 0.9
# total_time = 10

# временная сетка
h_min = np.min(np.diff(grid))
tau_W = h_min * initial_porosity / (2 * filtration_velocity) * (1 - EPS)
# tau_J = 2 * rho1 * M2**m / (k_a * M1 * initial_area_per_volume * rho2**m * (conc_left - c_eq)**m) / 50
kinetic_speed = k_a * np.power(conc_left * rho2 / M2, m)
Jo = initial_area_per_volume * kinetic_speed * alpha_rock
tau_J = rho1 * (1 - initial_porosity) / (Jo * alpha_rock) * (1 - EPS)
tau_J2 = conc_left * initial_porosity / \
    (2 * ( alpha_acid / rho1 * Jo + conc_left * filtration_velocity / h_min)) * (1 - EPS)
# tau = min(tau_W, tau_J)
tau = min(tau_W, tau_J, tau_J2)
# if tau_J < tau_W:
# 	print("Ограничитель по скорости реакции: порода")
# else:
# 	print("Ограничитель по скорости переноса")
	
if tau_J < tau_W and tau_J < tau_J2:
	print("Ограничитель по скорости реакции: порода")
elif tau_J2 < tau_W and tau_J2 < tau_J:
	print("Ограничитель по скорости реакции: кислота")
else:
	print("Ограничитель по скорости переноса")
t = np.arange(0, total_time + tau, tau)
num_of_time_points = t.shape[0]
print(f"{num_of_time_points} временных итераций")
# num_of_time_points = 500

# Начальные условия
initial_porosity_field = np.zeros(num_of_cell_points) + initial_porosity
initial_saturation_field = np.zeros(num_of_cell_points) + saturation_initial

# Граничные условия
# вектор скорости фильтрации
velocity_left = np.zeros(num_of_time_points) + filtration_velocity
# вектор водонасыщенности
water_saturation_left = np.zeros(num_of_time_points) + saturation_left
# вектор концентрации кислоты
acid_conc_left = np.zeros(num_of_time_points)
acid_conc_left[:]

acid_conc_left[:] = conc_left
# acid_conc_left[:] = 0
# for i, time in enumerate(t):
# 	if time < 20:
# 		acid_conc_left[i] = 0.3
# 	elif time < 40:
# 		acid_conc_left[i] = 0
# 		water_saturation_left[i] = 0
# 	elif time < 60:
# 		acid_conc_left[i] = 0.3
# 		water_saturation_left[i] = 1
# 	else:
# 		acid_conc_left[i] = 0
# 		water_saturation_left[i] = 0
```
