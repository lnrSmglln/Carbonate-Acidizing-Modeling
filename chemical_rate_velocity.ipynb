```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def a_v(poro, a_0, beta, poro0):
    return a_0 * ((1-poro) / (1 - poro0))**beta


def r_p(poro, r0, poro0, beta = 1):
    return r0 * (poro * ( 1 - poro0 ) / (poro0 * (1 - poro)))**(beta)
```

```python
poro0 = 0.2
r0 = 0.0005
beta = 1

poro = np.linspace(0, poro0, 100)
plt.plot(poro, r_p(poro, r0, poro0, beta))
```

```python
a_0 = 1
plt.plot(poro, a_v(poro, a_0, beta, poro0))
```
