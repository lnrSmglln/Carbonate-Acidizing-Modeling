```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def mu_cacl2(w, T):
    mu_water = 1e-3
    b1 = 3.4143
    b2 = -0.3062
    b3 = 84.004
    return mu_water * np.exp(w * ( b1  + b2 * 1e-2 * T - b3 * 1e-7 * T**2 ))

def mu_mgcl2(w, T):
    mu_water = 1e-3
    b1 = 5.0164
    b2 = -0.8807
    b3 = -96.036
    return mu_water * np.exp(w * ( b1  + b2 * 1e-2 * T - b3 * 1e-7 * T**2 ))

def mu_salts(w_cacl2, w_mgcl2, T):
    mu_water = 1e-3
    cacl2 = {
        "b1": 3.4143,
        "b2": -0.3062,
        "b3": 84.004,
    }
    mgcl2 = {
        "b1": 5.0164,
        "b2": -0.8807,
        "b3": -96.036
    }
    return mu_water * np.exp(
        w_cacl2 * ( cacl2["b1"]  + cacl2["b2"] * 1e-2 * T - cacl2["b3"] * 1e-7 * T**2 ) +
        w_mgcl2 * ( mgcl2["b1"]  + mgcl2["b2"] * 1e-2 * T - mgcl2["b3"] * 1e-7 * T**2 )
    )
```

```python
w = np.arange(0, 0.1, 0.01)
T = 20
mu_w_cacl2 = mu_cacl2(w, T)
mu_w_mgcl2 = mu_mgcl2(w, T)
mu_w_salts = mu_salts(w, w, T)
mu_w_cacl2, mu_w_mgcl2
```

```python
plt.plot(w, mu_w_cacl2, label=r"$\mu( w_{{CaCl}_2}, T=20^oC)$") 
plt.plot(w, mu_w_mgcl2, label=r"$\mu( w_{{MgCl}_2}, T=20^oC)$")
plt.plot(w, mu_w_salts, label=r"$\mu( w_{{CaCl}_2} + w_{{MgCl}_2}, T=20^oC)$")
plt.legend()
plt.show()
```
